## fibonacci(N, X)
La relación `fibonacci(N, X)` calcula el término número N de la sucesión de Fibonacci.  
La sucesión es: 0, 1, 1, 2, 3, 5, 8, ...

Casos base (cuando N es 0 o 1):

-fibonacci(0, 0) → el término 0 vale 0

-fibonacci(1, 1) → el término 1 vale 1

Regla recursiva:  
-Para N > 1, se cumple que f(n) = f(n-1) + f(n-2)


si N = 3, se cumple N>1 entoces:

N1 = 3-1 = 2 -> (2,x1)

N2 = 3-2 = 1 ->(1,x2)

 Para (2,x1)
 
 N1 = 2-1 =1 ->(1,x1) caso base
 
 N2 = 2-2 = 0-> (0,x2) caso base
 X1 = 1+0 = 1

 Para (1,x2) caso base = 1
 
 x= x1 +x2 = 2




```python
janus.consult('fibonacci(N, X)', '''
fibonacci(0, 0).
fibonacci(1, 1).
fibonacci(N, X) :- N > 1, N1 is N-1, fibonacci(N1, X1), N2 is N-2, fibonacci(N2, X2), X is X1+X2.
''')
list(janus.query('fibonacci(3, X).'))
```
